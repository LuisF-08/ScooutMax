# Dados Interpretados

## Informações Gerais da Partida

| Campo   | Descrição                                      |
|---------|------------------------------------------------|
| **Country** | País onde a liga/campeonato é disputado.      |
| **League**  | Nome da liga ou torneio (ex: Premier League, Brasileirão). |
| **Season**  | Ano ou temporada correspondente ao jogo.      |
| **Date**    | Data de realização da partida.                |
| **Time**    | Horário de início da partida.                 |
| **Home**    | Nome do time mandante (casa).                 |
| **Away**    | Nome do time visitante (fora).                |

---

## Placar e Resultado Final

| Campo | Descrição |
|-------|-----------|
| **HG** (Home Goals) | Número de gols marcados pelo time mandante. |
| **AG** (Away Goals) | Número de gols marcados pelo time visitante. |
| **Res** (Result)    | Resultado final do jogo:<br>• **H** = Vitória da casa (Home)<br>• **D** = Empate (Draw)<br>• **A** = Vitória do visitante (Away) |

---

## Odds de Apostas (Cotações de Fechamento)

> A sigla no final das odds indica o mercado:  
> **CH** = Casa (Closing Home) · **CD** = Empate (Closing Draw) · **CA** = Visitante (Closing Away)  
> O **C** indica que são **odds de fechamento** (as cotações finais antes de a partida começar).

### Pinnacle

| Campo  | Descrição |
|--------|-----------|
| **PSCH** | Odd de fechamento da Pinnacle para vitória do mandante |
| **PSCD** | Odd de fechamento da Pinnacle para empate |
| **PSCA** | Odd de fechamento da Pinnacle para vitória do visitante |

### Estatísticas Gerais do Mercado

| Campo | Descrição |
|-------|-----------|
| **MaxCH / MaxCD / MaxCA** | Maior odd disponível no mercado para Casa, Empate e Visitante, respectivamente |
| **AvgCH / AvgCD / AvgCA** | Média das odds de todas as casas de apostas comparadas |

### Betfair Exchange (BFE)

| Campo   | Descrição |
|---------|-----------|
| **BFECH** | Odd de fechamento na bolsa de apostas Betfair Exchange para vitória do mandante |
| **BFECD** | Odd de fechamento na Betfair Exchange para empate |
| **BFECA** | Odd de fechamento na Betfair Exchange para vitória do visitante |

### Bet365 (B365)

| Campo    | Descrição |
|----------|-----------|
| **B365CH** | Odd de fechamento na Bet365 para vitória do mandante |
| **B365CD** | Odd de fechamento na Bet365 para empate |
| **B365CA** | Odd de fechamento na Bet365 para vitória do visitante |

-NOTA: essa estatistica não tinha pois a BET não existia antes disso

```python
# Os nulos de Odd são muito altos e podem ser retirados dependendo das regras de negocio
df[df['HG'].isnull()]
```

```python
# O Jogo que está como nulo foi o do acidentwe do avião da chapecoense :(
# Uma forma de preencher isso é colocando 3 X 3 para ambos pois foi W.O

filtro = (
    df['Season'] == 2016) & (df['Home'] == 'Chapecoense-SC') & (df['HG'].isna())

df.loc[filtro, ['AG', 'HG']] = 3.0
df.loc[filtro, 'Res'] = 'D'

df[(df['Season'] == 2016) & (df['Home'] == 'Chapecoense-SC') & (df['Away'] == 'Atletico-MG')]
```

```python
# verificando duplicatas , nesse caso há zero duplicatas
df.duplicated().sum()
```

```python
# verificando os nulos na Pinnacle
# Filtra se houver nulo em QUALQUER uma das 3 colunas da Pinnacle
colunas_pinnacle = ["PSCH", "PSCD", "PSCA"]

df[df[colunas_pinnacle].isna().any(axis=1)]
# como houve uma fusão de empresas a pinnacle deixou de exitir e por isso não esta mais nesse .csv
```

```python
# Visão geral de Nulos (contagem absoluta e percentual)
nulos = df.isna().sum()
pct_nulos = (df.isna().sum() / len(df)) * 100
pd.DataFrame({'Nulos': nulos, '% Nulos': pct_nulos.round(2)})
```

```python
# 5. Resumo estatístico das Odds/Gols (para detectar valores absurdos ou negativos)
df.describe().T[['min', 'max','mean']]
```

```python

# Checar se a regra de resultado (Res) bate com os gols (HG x AG)
# Exemplo: Jogos com resultado 'H' onde o time da casa não fez mais gols
inconsistencias = df[
    ((df['Res'] == 'H') & (df['HG'] <= df['AG'])) |
    ((df['Res'] == 'A') & (df['AG'] <= df['HG'])) |
    ((df['Res'] == 'D') & (df['HG'] != df['AG']))
]
len(inconsistencias)
```
